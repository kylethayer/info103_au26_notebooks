# Demo of User Input

We can ask the user to type something using the `input` function

```python
user_input = input("Type something in here:")
```

```python
user_input = input("Type in your name, so I can greet you:")

print("Hello " + user_input + "! Nice to meet you!")
```

## Use input to choose options for actions

```python
user_input = input(
    "Choose your action: \n" + 
    " 1. list latest messages \n" +
    " 2. delete last message \n")

if user_input == "1":
    print("Pretending to print the latest messages:")
    print(" 1. I am angry at everyone")
    print(" 2. How are oyu all today")
    print()
elif user_input == "2":
    print("Pretending to delete the last message")
    print()
```

## Use a while loop to repeat actions until "exit"

```python
# temporarly save something in user_input, so we can start our loop
user_input = "temporary fake input"

while user_input != "exit":
    user_input = input(
        "Choose your action: \n" + 
        " 1. list latest messages \n" +
        " 2. delete last message \n" +
        " 3. block everyone who has messaged me \n" + 
        " Type 'exit' when you are done \n"    
    )
    
    if user_input == "1":
        print("Pretending to print the latest messages:")
        print(" 1. I am angry at everyone")
        print(" 2. How are oyu all today")
        print()
    elif user_input == "2":
        print("Pretending to delete the last message")
        print()
    elif user_input == "3":
        print("Pretending to block everyone who messaged me")
        print()
```
